### One-off cell to rename results

One-off cell to rename the `results` subfolder which were originally named `metrics`.

```python
# Rename `metrics` folders to `results`
from pathlib import Path

root = Path('..\\scenarios')

for scenario in root.iterdir():
    metrics_folder = scenario / 'metrics'
    results_folder = scenario / 'results'

    if metrics_folder.exists():
        metrics_folder.rename(results_folder)
        #print(f'Renamed: {metrics_folder} => {results_folder}')
        print('U', end='')
    else:
        print('-', end='')

print('\n\n-- Complete --')
```
