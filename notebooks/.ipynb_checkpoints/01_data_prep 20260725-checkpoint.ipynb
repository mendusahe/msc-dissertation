### Helper functions to examine a DataFrame (`summarise_column`, `show_summary`)

```python
def summarise_columns(df):

    summary = pd.DataFrame({
        'column': df.columns,
        'type': df.dtypes.astype(str).values,
        'n_vals': [df[col].nunique(dropna=False) for col in df.columns],   # Include NA in unique count
        'n_missing': df.isna().sum().values
    })
    summary.set_index('column', inplace=True)
    
    samples = [df[col].unique().tolist()[:10] for col in df.columns]
    #for col in df.columns:
    #    print(df[col].unique().tolist()[:10])
    summary['samples'] = samples
    return summary

def show_summary(summary):
    print(f'{"Column":30} {"dtype":12}{"N Unique":8}{"N Missing":30}{"Samples":10}')
    print(f'{'-'*100}')
    for col, row in summary.iterrows():
        print(f'{col:30}{row.type:12}{row.n_vals:8d}{row.n_missing:8d}{row.samples[:10]}')
```

**Observations**  
The data is relatively tid but requires several transformation:  
* int columns with 2 values - No transformation required.  
* object columns with 2 values - Convert to int. These will be treated as binary.  
* object columns with >2 values - Convert to categorical.  
* int columns with >2 values - Examine the data and sematics and convert to categorical is appropriate.


Missing values:
* categorical columns - replace `nan` with "missing". 
* int and float columns - Do not imupte values in the raw data to avoid leaking data from later test sets into the training set. Instead, add columns to indicate where missing values existe.

><font color=#800000>**Explanation of Code**
>
><font color=#800000>This cell replaces missing values according to the following rules:  
> * <font color=#800000>Categorical features: replace with 'missing'.  
> *  <font color=#800000>Numerical features with <10% (3000 approximately) values missing: replace with median.  
> *  <font color=#800000>Numerical features with >10% values missing: replace with median and add column 'was_missing'.
> *  <font color=#800000>String features: replace with 'missing'
> *  <font color=#800000>Where missing values in columns are closely correlated, only add one 'waw_missing' column per group of features. To populated the 'was_missing' column for a group, take the majority vote.

```python
# Replace the missing string values
string_cols = summary_missing[summary_missing['dtype']=='string']['column'].to_list()

# Replace missing values for string features with "missing"
df_1 = df[string_cols].fillna("missing")

# Replace missing values for numeric feature with median value
num_cols = summary_missing[summary_missing['dtype']=='float64']['column'].to_list()
df_1[num_cols] = df[num_cols].apply(lambda col: col.fillna(col.median()))

# Checck that no missing values remain
print('df_1.shape:', df_1.shape)
df_1.isna().sum().values
```

```python
# Populate 'was_missing' columns for each group of correlated numerical featrues which had
# missing values by taking the majority vote
missing_num_groups = {
    'AMT_REQ_CREDIT_BUREAU_MISSING': [
        'AMT_REQ_CREDIT_BUREAU_DAY', 'AMT_REQ_CREDIT_BUREAU_HOUR', 'AMT_REQ_CREDIT_BUREAU_MON', 
        'AMT_REQ_CREDIT_BUREAU_QRT', 'AMT_REQ_CREDIT_BUREAU_WEEK', 'AMT_REQ_CREDIT_BUREAU_YEAR'],
    'EXT_SOURCE_1_MISSING': ['EXT_SOURCE_1'],
    'EXT_SOURCE_3_MISSING': ['EXT_SOURCE_3'],
    'OWN_CAR_AGE_MISSING': ['OWN_CAR_AGE'],
    'STATS_MISSING': ['TOTALAREA_MODE','YEARS_BEGINEXPLUATATION_AVG','YEARS_BEGINEXPLUATATION_MEDI',
                      'YEARS_BEGINEXPLUATATION_MODE','FLOORSMAX_AVG','FLOORSMAX_MEDI','FLOORSMAX_MODE',
                      'LIVINGAREA_AVG','LIVINGAREA_MEDI','LIVINGAREA_MODE','ENTRANCES_AVG','ENTRANCES_MEDI',
                      'ENTRANCES_MODE','APARTMENTS_AVG','APARTMENTS_MEDI','APARTMENTS_MODE','ELEVATORS_AVG',
                      'ELEVATORS_MEDI','ELEVATORS_MODE','NONLIVINGAREA_AVG','NONLIVINGAREA_MEDI','NONLIVINGAREA_MODE',
                      'BASEMENTAREA_AVG','BASEMENTAREA_MEDI','BASEMENTAREA_MODE','LANDAREA_AVG','LANDAREA_MEDI',
                      'LANDAREA_MODE','YEARS_BUILD_AVG','YEARS_BUILD_MEDI','YEARS_BUILD_MODE','FLOORSMIN_AVG',
                      'FLOORSMIN_MEDI','FLOORSMIN_MODE','LIVINGAPARTMENTS_AVG','LIVINGAPARTMENTS_MEDI',
                      'LIVINGAPARTMENTS_MODE','NONLIVINGAPARTMENTS_AVG','NONLIVINGAPARTMENTS_MEDI',
                      'NONLIVINGAPARTMENTS_MODE','COMMONAREA_AVG','COMMONAREA_MEDI','COMMONAREA_MODE']
}
for key, value in missing_num_groups.items():
    print(key, 'Value count:', len(value))
    print(value)
```

```python
# Populate the 'was_missing' columns based on majority of columns missing in the source data
for missing_col, group in missing_num_groups.items():
    majority = len(group) // 2 +1
    print('Populating column:', missing_col, 'with majority', majority)
    df_1[missing_col] = (df[group].isna().sum(axis=1) >= majority).astype(int)
```

### Add WAS_MISSING columns

><font color=#800000>**Explanation of Code**
>
><font color=#800000>Missing values carry information that is useful to the classifier, so columns are added to indicate where "missing" has replace NA. Because many categorical columns had missing values, this would lead to a large number of "WAS_MISSING" columns being added. To reduce this number, single indicator columns are added to represent closely correlated groups of columns. Where the column are not perfectly correltated, a majority vote is taken amongst the column to derive the value of the indicator.

```python
# Populate 'WAS_MISSING' columns for each group of correlated numerical featrues which had
# missing values by taking the majority vote
missing_num_groups = {
    'AMT_REQ_CREDIT_BUREAU_MISSING': [
        'AMT_REQ_CREDIT_BUREAU_DAY', 'AMT_REQ_CREDIT_BUREAU_HOUR', 'AMT_REQ_CREDIT_BUREAU_MON', 
        'AMT_REQ_CREDIT_BUREAU_QRT', 'AMT_REQ_CREDIT_BUREAU_WEEK', 'AMT_REQ_CREDIT_BUREAU_YEAR'],
    'EXT_SOURCE_1_MISSING': ['EXT_SOURCE_1'],
    'EXT_SOURCE_3_MISSING': ['EXT_SOURCE_3'],
    'OWN_CAR_AGE_MISSING': ['OWN_CAR_AGE'],
    'STATS_MISSING': ['TOTALAREA_MODE','YEARS_BEGINEXPLUATATION_AVG','YEARS_BEGINEXPLUATATION_MEDI',
                      'YEARS_BEGINEXPLUATATION_MODE','FLOORSMAX_AVG','FLOORSMAX_MEDI','FLOORSMAX_MODE',
                      'LIVINGAREA_AVG','LIVINGAREA_MEDI','LIVINGAREA_MODE','ENTRANCES_AVG','ENTRANCES_MEDI',
                      'ENTRANCES_MODE','APARTMENTS_AVG','APARTMENTS_MEDI','APARTMENTS_MODE','ELEVATORS_AVG',
                      'ELEVATORS_MEDI','ELEVATORS_MODE','NONLIVINGAREA_AVG','NONLIVINGAREA_MEDI','NONLIVINGAREA_MODE',
                      'BASEMENTAREA_AVG','BASEMENTAREA_MEDI','BASEMENTAREA_MODE','LANDAREA_AVG','LANDAREA_MEDI',
                      'LANDAREA_MODE','YEARS_BUILD_AVG','YEARS_BUILD_MEDI','YEARS_BUILD_MODE','FLOORSMIN_AVG',
                      'FLOORSMIN_MEDI','FLOORSMIN_MODE','LIVINGAPARTMENTS_AVG','LIVINGAPARTMENTS_MEDI',
                      'LIVINGAPARTMENTS_MODE','NONLIVINGAPARTMENTS_AVG','NONLIVINGAPARTMENTS_MEDI',
                      'NONLIVINGAPARTMENTS_MODE','COMMONAREA_AVG','COMMONAREA_MEDI','COMMONAREA_MODE']
}
for key, value in missing_num_groups.items():
    print(key, 'Value count:', len(value))
    print(value)
```

```python
# Populate the 'WAS_MISSING' columns based on majority of columns being missing in the source data

df6 = df5.copy(deep=False)

# For each group of columns, count the number which were missing in the original Dataset and 
# set WAS_MISSING=1 if the number exceeds the number of columns / 2
for missing_col, group in missing_num_groups.items():
    majority = len(group) // 2 +1
    print('Populating column:', missing_col, 'with majority', majority)
    df6[missing_col] = (df[group].isna().sum(axis=1) >= majority).astype(int)
    #df_6[missing_col] = (df5[group].sum(axis=1) >= majority).astype(int)
```

 ## Imputer (NOT REQUIRED)

To avoid information leakage from the test set intoe the training set, a data cleansing process must first be trained on the training set then applied to both the train and test sets. Because this study requires training many models across multiple scenairos using cross validtaion, an imputer function is required which can be incorporated into the training pipeline. This section implements a custom imputer function based on scikitlearn's 

```python
class CustomImputer(BaseEstimator, TransformerMixin):
    def fit(self, X, y=None):
        # compute medians
        # compute majority-vote rules
        # store them
        return self
    
    def transform(self, X):
        # apply stored medians
        # apply stored majority-vote rules
        return X
```

```python
# Read the training set
X_train = pd.read_parquet('..\\data\\processed\\X_train.parquet', engine='pyarrow')
print(X_train.shape)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
import joblib

# Get a list of numeric column which are missing values, then create a ColumnTransformer
# to transform only those columns
num_cols_with_missing = summary_missing[summary_missing['dtype']=='float64']['column'].to_list()
print('Creating median_imputer to checks', len(num_cols_with_missing),'numeric columns')

median_imputer = ColumnTransformer(
    transformers=[
        ('median_imputer', SimpleImputer(strategy='median'), num_cols_with_missing)
    ],
    remainder='passthrough'
)
joblib.dump(median_imputer, '..\\models\\xgboost\\median_imputer.pkl')
```

```python
# Test on the testset
median_imputer.fit(X_train)
X_test_imputed = median_imputer.transform(X_test)

X_test_imputed_df = pd.DataFrame(X_test_imputed, columns=X_test.columns, index=X_test.index)

for col in num_cols_with_missing[:5]:
    missing_indices = X_test[X_test[col].isna()].index
    print('----Column', col, 'has', len(missing_indices), 'missing rows----')
    print(X_test.loc[missing_indices[:1], col])
    print(X_test_imputed_df.loc[missing_indices[:1], col])
```

## GAN runtime testing

### Create GAN runtime scenarios

```python
import numpy as np

# Create four different-sized scenarios
n_samples = [300, 1500, 7000, 19000]

minority_df = train_df[train_df['target']==1]
print('minority shape:', minority_df.shape)

for n in n_samples:
    scenario_df = minority_df.sample(n, random_state=1).index
    name = scenario_name(n, None, 'rt-minority')
    path = f'..\\data\\scenarios\\{name}.txt'
    np.savetxt(path, scenario_df, fmt='%d')
    print(f'{name}: {n:d}')    
```

### Measure GAN runtimes

```python
import json, time
from sdv.single_table import CTGANSynthesizer

# Instantiate the GAN once
synthesizer = CTGANSynthesizer(
    metadata,
    generator_dim = (64, 64),
    discriminator_dim = (64, 64),
    epochs=400,
    verbose=True)

# Iterate through the GANs
rt_log = {}
for n in n_samples:
    print(f'Training GAN on {n:d} samples...')
    start = time.perf_counter()

    name = scenario_name(n, None, 'rt-minority')
    path = f'..\\data\\scenarios\\{name}.txt'
    
    # Create the dataset
    idx = pd.read_csv(path, header=None)[0]
    rt_scenario = train_df.loc[idx]
   
    #Train the GAN
    synthesizer.fit(rt_scenario)
    
    end = time.perf_counter()
    elapsed_time = (end - start)/60
    print(f'Training completed in {elapsed_time:.1f} minutes')

    # Record the result
    runlog = {'size': n, 'runtime': elapsed_time//60}
    print('run', runlog)
    rt_log[f'Sample {n:d}'] = runlog

with open('..\\models\\gan\\gan-runtimes.json', 'w') as f:
    json.dump(rt_log, f, indent=4)
```
