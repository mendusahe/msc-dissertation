**Purpose of this notebook**  
This notebook creates metadata that is used by all CTGAN models, based on the training dataset.

**NOTES**  
CTGAN must be trained on imputed training data. It is not correct to train it on the raw training data (with missing numerical values) and then add the median values to the synthetic data.

### Load the training set

```python
import pandas as pd

import pandas as pd

# Read the training set. Convert the label DataFrames back to Series
X_train = pd.read_parquet('..\\data\\processed\\X_train.parquet', engine='pyarrow')
y_train = pd.read_parquet('..\\data\\processed\\y_train.parquet', engine='pyarrow')['TARGET']

print(X_train.shape, y_train.shape)
```

```python
X_train
```

## Impute missing values (`hf_preprocess`)

```python
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer

# Create a preprocessor to replace missing with median for numeric columns only. 
numeric_cols = X_train.select_dtypes(include=['float64']).columns
hf_preprocess = ColumnTransformer(
    transformers=[('median_imputer', SimpleImputer(strategy='median'), numeric_cols)],
    remainder='passthrough',
    verbose_feature_names_out=False
).set_output(transform='pandas')
```

## Detect metadata from DataFrame

```python
from pathlib import Path
from sdv.metadata import Metadata

# Create and save metadata for CTGAN based on the training data.
X_train_imp = hf_preprocess.fit_transform(X_train)

# Instantiate the GAN metadata
#metadata_1 = Metadata()
metadata_1 = Metadata.detect_from_dataframe(X_train_imp, table_name='credit')

# Save a copy of metadata
metadata_1.save_to_json(root+'metadata_1.json', mode='overwrite')
```

## Compare the DataFrame and CTGAN metadata

```python
import pandas as pd

# Create a DataFrame summarising the training dataset before imputing missing values
X_train_summary = pd.DataFrame({
    'column': X_train.columns,
    'dtype': X_train.dtypes.astype(str).values,
    'n_vals': [X_train[col].dropna().nunique() for col in X_train.columns],
    'n_missing': X_train.isna().sum().values
})
X_train_summary.sort_values(['column'], inplace=True)
X_train_summary.set_index("column", inplace=True)
#print(X_train_summary.head(2))

# Create a DataFrame summarising the imputed training dataset
X_train_imp_summary = pd.DataFrame({
    'column': X_train_imp.columns,
    'dtype': X_train_imp.dtypes.astype(str).values,
    'n_vals': [X_train_imp[col].dropna().nunique() for col in X_train_imp.columns],
    'n_missing': X_train_imp.isna().sum().values
})
X_train_imp_summary.sort_values(['column'], inplace=True)
X_train_imp_summary.set_index("column", inplace=True)
#print(dframe_summary.head(3))
```

```python
def hf_compare_df_and_metadata(summary_df, metadata):
    '''Compare the datatypes in a DataFrame and a Metadata object.'''

    # Create a dict of metadata column sdtypes then compare the DataFrane and metadata
    metadata_columns = metadata.to_dict()['tables']['credit']['columns']
    sdtypes = {col: val['sdtype'] for (col, val) in metadata_columns.items()}
    
    print(f"{'Column':33} {'dtype':10} {'missing':8} {'n_vals':8} {'df dtype':10} | {'sdtype':10}")
    print('=======================================================================================')
    for col, row in summary_df.iterrows():
        #col = row.column
        X_row = summary_df.loc[col]
        #print(X_row['dtype'])
        print(f'{col:33} {X_row['dtype']:10} {X_row.n_missing:8d} {row.n_vals:8d} {row['dtype']:10} | {sdtypes[col]:10}')
```

## CTGAN helper functions OLD CODE

### Read scenario indices and create minotiry training sample (`hf_read_scenario_data`)

```python
from pathlib import Path

from pathlib import Path
import numpy as np

def hf_read_scenario_data(tier, ir, bootstrap):
    '''Get the training data for a scenario, including only minority class samples.
    
    Summary
    -------
    Generate the path to the indices from the input parameters. Read the indices then select
    the training data from the full training, filter on positive samples and return the result.
    '''
    # Get the path to the indices
    scenario = f'{tier}-ir{ir:03d}-b{bootstrap:02}'
    root = Path('..') / 'scenarios' / scenario

    # Read the indices for the scenario
    train_file = f'training-{scenario}.txt'
    train_path = root / train_file
    train_indices = np.loadtxt(train_path, dtype=int)

    # Use the indices to filter the training set, then further filter on positive samples
    y_scenario = y_train.loc[train_indices]
    X_scenario = X_train.loc[train_indices]
    
    return  X_scenario[y_scenario==1]
```

```python
from time import perf_counter

def train_ctgan_model(tier, ir, bootstrap):
    '''
    Train a CTGAN model using minority class samples only from a predefined training set for a
    scenario.

    NOTES
    -----
    CTGAN models are not saved. Data synthesis must be performed while the model remains in memory.
    '''
    hf_check_dependencies()

    # Read the scenario training set
    X_minority_raw = hf_read_scenario_data(tier, ir, bootstrap)

    # Imput the missing values in the training set
    X_minority_imp = hf_preprocess.fit_transform(X_minority_raw)
    #print(X_minority_imp.shape)

    # Load the metadata for CTGAN
    metadata = hf_ctgan_metadata(X_minority_imp)

    
    return None

    
```
