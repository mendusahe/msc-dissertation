**Purpose of this notebook**  
Create JSON files containing the tuned hyperparameters used to train XGBoost for each scenario. For this study, hyperparameters and metadata are stored in an Excel spreadsheet. This notebook reads the hyperparameters for each scenario from the spreadsheet and saves them as JSON files.
```
..
    |-- models
        |-- xgboost
            |-- hyperparameters
                |-- hyperparameters.xlsx
                |-- large-ir03.json
                |-- large-ir09.json
                |-- ...
                |-- small-ir81.json
```

## Read hyperparameters from Excel and save as JSON files

This cell can be skipped if hyperparameter JSON files are produced programatically.

````python
import os, json
from pathlib import Path
import pandas as pd
import numpy as np

# Spreadsheet name, column names and types
hyperparameter_ss = "hyperparameters.xlsx"
param_dtypes = {
    'colsample_bytree': np.float64, 
    'learning_rate': np.float64,
    'reg_alpha': np.float64, 
    'reg_lambda': np.float64,
    'subsample': np.float64,
    'scale_pos_weight': np.float64,
    'min_child_weight': np.int32,
    'max_depth': np.int32,
    'gamma': np.int32
}
param_cols = ['scenario'] + [c for c in param_dtypes.keys()]

# Hyperparameter path
root = Path('..')
hyperparameter_path = root / 'models' /  'xgboost' / 'hyperparameters' 

# Load spreadsheet using the specified column names and dtypes, and scenario as the index.
params_df = pd.read_excel(
    hyperparameter_path / hyperparameter_ss,
    sheet_name='params',
    usecols=param_cols,
    dtype=param_dtypes,
    index_col='scenario')
#print(params_df.head(2))
#print(params_df.info())

# Iterate through scenarios and save as JSON files. 
# Use named tuples so that the index, column names and dtypes are retained
for n, row in enumerate(params_df.itertuples(name='params')):

    # Convert row to dict, read the index (scenario) then remove it from the dict
    params = row._asdict()
    scenario = params['Index']
    params.pop('Index')
    
    # Create the output file path from the scenario and write the dict to JSON
    json_path = hyperparameter_path / (scenario + '.json')
    print('Writing file:', json_path)
    with open(json_path, "w") as f:
        json.dump(params, f, indent=4)

print(f"\n{n+1:d} Hyperparameter JSON files created successfully.")
````
