**Purpose of this notebook**  
This notebook creates metadata that is used by all CTGAN models, based on the training dataset. Metadata is saved in the folder below and read during CTGAN training.
```
..
    |-- models
        |-- ctgan
            |-- ctgan_metadata.json
```

## Load the training set

````python
import pandas as pd

# Read the training set. Convert the label DataFrames back to Series
X_train = pd.read_parquet('..\\data\\processed\\X_train.parquet', engine='pyarrow')
y_train = pd.read_parquet('..\\data\\processed\\y_train.parquet', engine='pyarrow')['TARGET']

print(X_train.shape, y_train.shape)
X_train.info()
````

## Process the training pool to derive CTGAN metadata

### Impute missing values (`hf_preprocess`)

The imputed data is not saved, and is created only to ensure that column `sdtypes` are created predictably.

````python
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer

# Create a preprocessor to replace missing with median for numeric columns only. 
numeric_cols = X_train.select_dtypes(include=['float64']).columns
hf_preprocess = ColumnTransformer(
    transformers=[('median_imputer', SimpleImputer(strategy='median'), numeric_cols)],
    remainder='passthrough',
    verbose_feature_names_out=False
).set_output(transform='pandas')
````

### Detect metadata from DataFrame

Metadata is initially detected automatically by `Metadata`.

````python
from pathlib import Path
from sdv.metadata import Metadata

# Create and save metadata for CTGAN based on the training data.
X_train_imp = hf_preprocess.fit_transform(X_train)

# Instantiate the GAN metadata
metadata_1 = Metadata.detect_from_dataframe(X_train_imp, table_name='credit')

# Save a copy of metadata
metadata_1.save_to_json('..\\models\\ctgan\\metadata_1.json', mode='overwrite')
````

### Compare the DataFrame and CTGAN metadata

````python
import pandas as pd

# Create a DataFrame summarising the training dataset before imputing missing values
X_train_summary = pd.DataFrame({
    'column': X_train.columns,
    'type': X_train.dtypes.astype(str).values,
    'n_vals': [X_train[col].dropna().nunique() for col in X_train.columns],
    'n_missing': X_train.isna().sum().values
})
X_train_summary.sort_values(['column'], inplace=True)
X_train_summary.set_index("column", inplace=True)
#print(X_train_summary.head(2))

# Create a DataFrame summarising the imputed training dataset
X_train_imp_summary = pd.DataFrame({
    'column': X_train_imp.columns,
    'type': X_train_imp.dtypes.astype(str).values,
    'n_vals': [X_train_imp[col].dropna().nunique() for col in X_train_imp.columns],
    'n_missing': X_train_imp.isna().sum().values
})
X_train_imp_summary.sort_values(['column'], inplace=True)
X_train_imp_summary.set_index("column", inplace=True)
#print(dframe_summary.head(3))
````

````python
def hf_compare_df_and_metadata(summary_df, metadata):
    '''Compare the datatypes in a DataFrame and a Metadata object.'''

    # Create a dict of metadata column sdtypes then compare the DataFrane and metadata
    metadata_columns = metadata.to_dict()['tables']['credit']['columns']
    sdtypes = {col: val['sdtype'] for (col, val) in metadata_columns.items()}
    
    print(f"{'Column':33} {'dtype':10} {'missing':8} {'n_vals':8} {'df dtype':10} | {'sdtype':10}")
    print('=======================================================================================')
    for col, row in summary_df.iterrows():
        #col = row.column
        X_row = summary_df.loc[col]
        #print(X_row['dtype'])
        print(f'{col:33} {X_row['type']:10} {X_row.n_missing:8d} {row.n_vals:8d} {row['type']:10} | {sdtypes[col]:10}')
````

````python
hf_compare_df_and_metadata(X_train_imp_summary, metadata_1)
````

**Observations**  
`detect_from_dataframe` has made numerous assumptions about the data and selected inappropriate datatypes:
* Several columns changed from numerical (float64 and int64) or Boolean to categorical.
* Some int64 columns with only two values should be categorical. Confirm that CTGAN has made all those categorical.
* It includes email, city, and id columns. These must be updated to boolean, numeric, or categorical

The following transformations are required from the DataFrame to CTGAN Metadata.
* int64, n=2: Convert to Boolean. Metadata=Boolean
* int64, n>2: int64 (no change). Metadata=Numerical, represent as integer
* float64: float64 (no change). Metadata=Numerical
* categorical: Convert to object. Metadata=Categorical

````python
import pandas as pd

# Iterate through the columns and update the metadata based on the DataFrame dtype
for col, row in X_train_imp_summary.iterrows():
    df_type = row.type
    print('Updating column:', col, '...   ', end='')
    if df_type == 'float64':
        metadata_1.update_column(col, sdtype='numerical', computer_representation='Float')
        print('float64')
    elif df_type == 'int64':
        metadata_1.update_column(col, sdtype='numerical', computer_representation='Int64')
        print('Int64')
    elif df_type == 'boolean':
        metadata_1.update_column(col, sdtype='boolean')
        print('boolean')
    elif df_type == 'category':
        metadata_1.update_column(col, sdtype='categorical')
        print('categorical')
    else:
        print('ERROR Unexpected data type')
        
print('\n-- Updated complete --')
````

````python
import json

# View the JSON dump of the metadata
print(json.dumps(metadata_1.to_dict(), indent=4))
````

````python
# Final comparison of the metadata with the DataFrame
hf_compare_df_and_metadata(X_train_imp_summary, metadata_1)
````

## Save the metadata

````python
# Save the final version of the metadata
metadata_1.save_to_json(r'..\\models\\ctgan\\ctgan_metadata.json', mode='overwrite')
````
