```python
import xgboost as xgb
#from xgboost import XGBClassifier, DMatrix, 
from sklearn.model_selection import GridSearchCV
from sklearn.metrics import make_scorer, f1_score
from sklearn.model_selection import train_test_split
import time

# Create xgb data objects
# 'enable_categorical' ensure categorical features are interpreted correctly
dtrain = xgb.DMatrix(X_poc, label=y_poc, enable_categorical=True)
dtest = xgb.DMatrix(X_test, label=y_test, enable_categorical=True)

params = {
    'objective': 'binary:logistic',     # Fixed
    'tree_method': 'hist',              # Fixed
    'nthread': 4,                       # Fixed
    'eval_metric': ['logloss', 'auc', 'aucpr'] # Fixed

    # Learning Rate
    'learning_rate': 0.05,              # Optional [0.03, 0.05, 0.1]
    # Sampling
    'subsample': 0.8,                   # [0.6, 0.8, 1.0] 0.6-1.0
    'colsample_bytree': 0.8,            # [0.6, 0.8, 1.0]0.6-1.0
    # Model complexity/Tree Structure
    'min_child_weight': 1               # [1, 3, 5] 3-10
    'max_depth': 6                     # [4, 6, 8] (reg)3-10
    # Regularisation
    #'gamma': xxx,                      # [0, 1, 5] (L2 reg) 0-5
    #'reg_alpha': xxx,                      # [0, 0.5, 1] (L1 reg)0-1
    #'reg_lambda': xxx                      # [0.5, 1, 2] 0-1
    # Imbalance
    #'scale_pos_weight': 26             # Maj/Min [x0.5, x1, x2]
} 
'''
Also, given early stopping: num_boost_round
1500 - 80k
1200 - 40k
800 - 20k
Song - 100-1000
Early stopping rounds: 80, 60, 40
'''

bst = xgb.train(params, dtrain, num_boost_round=300)
```

```python
from sklearn.metrics import roc_auc_score, f1_score

y_test_prob = bst.predict(dtest)
y_test_pred = (y_test_prob > 0.5).astype(int)

print('AUC:', roc_auc_score(y_test, y_test_prob))
print('F1:', f1_score(y_test, y_test_pred))
```

```python
# Sweep thresholds to find the optimum F1 score
best_f1 = 0
best_thresh = 0

for t in [i/100 for i in range(1,100)]:
    y_pred = (y_test_prob > t).astype(int)
    f = f1_score(y_test, y_pred)
    if f > best_f1:
        best_f1 = f
        best_thresh = t

print("Best threshold:", best_thresh)
print("Best F1:", best_f1)
```
