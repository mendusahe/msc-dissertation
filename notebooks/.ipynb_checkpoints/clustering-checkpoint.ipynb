````python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder, StandardScaler

def fit_preprocessor(train_df, cat_vars, num_vars, bin_vars):
    '''
    Train an encoder and apply to the training data, then train a scaler. The encoding must
    be applied so that one-hot scaling can be learned.
    '''
    # Create and train the encoder, then combine the binary and new ohe columns
    encoder = OneHotEncoder(handle_unknown='ignore', sparse_output=False)
    train_ohe_df = encoder.fit_transform(train_df[cat_vars])
    ohe_cols = encoder.get_feature_names_out(cat_vars)

    #dummy_vars = bin_vars + list(ohe_cols)
    ohe_vars = list(ohe_cols)

    # Create and train the numeric scaler, but do not transform the training set
    scaler_num = StandardScaler()
    scaler_num.fit(train_df[num_vars])

    # Create and train the binary scaler, but do not transform the training set
    scaler_bin = StandardScaler()
    scaler_bin.fit(train_df[bin_vars].astype(float))

    # Create and train the ohe scaler, but do not transform the training set
    scaler_ohe = StandardScaler()
    scaler_ohe.fit(pd.DataFrame(train_ohe_df, columns=ohe_cols, index=train_df.index))
    
    if False:
        # Create and train the scaler for all column (original code)
        scaler_dummy = StandardScaler()
        scaler_dummy.fit(
            pd.concat([
                train_df[bin_vars].astype(float),
                pd.DataFrame(train_ohe_df, columns=ohe_cols, index=train_df.index)
            ], 
            axis=1)
        )

    return encoder, scaler_num, scaler_bin, scaler_ohe, ohe_cols
````

````python
# Fit the preprocessor to the cluster train set
encoder, scaler_num, scaler_bin, scaler_ohe, ohe_cols = fit_preprocessor(
    cluster_train_df, cat_vars, num_vars, bin_vars)
print('Complete')
````

````python
def apply_preprocessors(df, encoder, scaler_num, scaler_bin, scaler_ohe, num_vars, bin_vars, cat_vars, ohe_cols, scale_bin=True, scale_ohe=False):

    # Encode categoricals
    ohe_raw_df = pd.DataFrame(
        encoder.transform(df[cat_vars]),
        columns=ohe_cols,
        index=df.index)

    # Scale numeric
    num_scaled_df = pd.DataFrame(
        scaler_num.transform(df[num_vars]),
        columns=num_vars,
        index=df.index)

    # Scale binary
    if scale_bin:
        bin_scaled_df = pd.DataFrame(
            scaler_bin.transform(df[bin_vars]),
            columns=bin_vars,
            index=df.index)
    else:
        bin_scaled_df = df[bin_vars]

    #Scale ohe
    if scale_ohe:
        ohe_scaled_df = pd.DataFrame(
            scaler_ohe.transform(ohe_raw_df),
            columns=ohe_raw_df.columns,
            index=df.index)
    else:
        ohe_scaled_df = ohe_raw_df

    #Scale dummy (original code)
    if False:
        dummy_raw_df = pd.concat(
            [df[bin_vars].astype(float), ohe_df],
             axis=1)
        if scale_dummy:
            dummy_scaled_df = pd.DataFrame(
                scaler_dummy.transform(dummy_raw_df),
                columns=dummy_raw_df.columns,
                index=df.index)
        else:
            dummy_scaled_df = dummy_raw_df

        # Combine the numeric and ohe columns
        combined_df = pd.concat([num_scaled_df, dummy_scaled_df], axis=1)

    # Combine the numeric, binary, and ohe columns
    combined_df = pd.concat([num_scaled_df, bin_scaled_df, ohe_scaled_df], axis=1)
    
    return combined_df
````

## PCA Dimension Reduction

````python
from sklearn.decomposition import PCA

def cluster_pca(df, variance=0.90):
    '''
    Apply PCA to a dataset.

    Parameters
    ----------
    X : pd.DataFrame
        Preprocessed dataset (scaled + encoded).
    variance : float
        Target explained variance for PCA (default 0.90).
    '''

    pca_model = PCA(n_components=variance, svd_solver='full')
    reduced_array = pca_model.fit_transform(df.values)

    # Convert the ndarray into a DataFrame with the original index
    reduced_df = pd.DataFrame(reduced_array, index=df.index)

    return reduced_df, pca_model
````

````python
# Perform PCA on the datasets (Approx runtime: 8 sec)
start = time.perf_counter()
encoded_train_maj_pca, pca_model = cluster_pca(encoded_train_maj_df, 0.9)
encoded_train_min_pca, pca_model = cluster_pca(encoded_train_min_df, 0.9)
end = time.perf_counter()
runtime = end - start
print(f'Runtime: {end-start:.0f}')
````

````python
import time

def hdbscan_cluster(df, min_cluster_size, min_samples):
    #min_cluster_size = 250
    #min_samples = 50
    size = df.shape[0]
    
    clusterer = HDBSCAN(
        min_cluster_size=min_cluster_size,
        min_samples=min_samples,
        metric='euclidean',
        cluster_selection_method='eom',
        copy=True, 
        n_jobs=4
    )
    #hdbpoc_df = encoded_train_maj_df.sample(size, random_state=42)
    
    # Fit the model and record the time
    start = time.perf_counter()
    clusterer.fit(df)
    end = time.perf_counter()
    runtime = end-start
    
    # Calculate simple statistics about the cluster
    labels = clusterer.labels_
    values, counts = np.unique(labels[labels!=-1], return_counts=True)
    print(f'Sample size: {size:d} min_cluster_size: {min_cluster_size:d} min_samples: {min_samples:d}  ')
    print(f'Runtime: {runtime:.2f} seconds ({runtime/60:.2f} minutes)  ')
    if counts.size==0:
        print('No clusters found')
    else:
        noise = np.size(labels[labels==-1])
        n_clusters = np.size(values)
        max_size = np.max(counts)
        min_size = np.min(counts)
        mean_size = np.mean(counts)
        noise_pct = (noise*100)//size
        print(f'Noise: {noise}({noise_pct}%) Clusters: {n_clusters:d} Min: {min_size} Max: {max_size} Mean: {mean_size:.2f}  ')
    return clusterer
````

### K-means sample generation

A function is created for perform to cluster a sample then save indices for stratified samples.

````python
from sklearn.cluster import MiniBatchKMeans
from sklearn.model_selection import train_test_split
from sklearn.metrics import silhouette_score

def create_kmeans_samples(df, cls, k, n_samples, sample_size, path, batch_size=1024, 
                          generate_samples=True, calc_metrics=False):
    '''
    Train k-means clusters and create stratified samples
    Parameters
    ----------
    df : pd.DataFrame
        Preprocessed dataset (scaled + encoded).
    cls : str
        Class of the dataset (maj or min).
    k : int
        Number of clusters.
    n_samples : int
        Number of stratified samples to generate.
    sample_size : int
        Size of stratified samples to generate.
    path : str
        Path to save the sample indices.
    batch_size : int
        Batch size 
    generate_samples, calc_metrics : Boolean
        Determine whether samples are generated and metrics are calculated.
    '''
    kmeans = MiniBatchKMeans(
        n_clusters=k,
        init='k-means++',
        max_iter=300,
        batch_size=batch_size,
        random_state=42,
        compute_labels=True,
        verbose=False,
        max_no_improvement=10,
        init_size=None
    )
    # Train k-means
    print(f'Training kmeans with class: {cls} and k: {k:d}... ', end='')
    start = time.perf_counter()
    kmeans.fit(df)
    end = time.perf_counter()
    print(f'Complete. Runtime: {end-start:.2f}')

    # Calcualte Metrics
    cluster_metrics = {}
    if calc_metrics:
        # To compute silhouette in reasonable time, operate on a random subsample
        idx_rnd = np.random.choice(len(df), size=1000, replace=False)
        df_sample = df.iloc[idx_rnd]
        label_sample = np.asarray(kmeans.labels_)[idx_rnd]
        cluster_metrics['silhouette_score']=silhouette_score(df_sample, label_sample)
        
        cluster_metrics['inertia']=int(kmeans.inertia_)
        counts = pd.Series(kmeans.labels_).value_counts().sort_index()
        cluster_metrics['cluster_freq']=counts.to_list()
        cluster_metrics['cluster_min']=int(np.min(counts))
        cluster_metrics['cluster_max']=int(np.max(counts))
        cluster_metrics['cluster_std']=int(np.std(counts))

    # Generate samples
    if generate_samples:
        for idx in range(n_samples):
            _, sample = train_test_split(
                df,
                test_size=sample_size,
                stratify=kmeans.labels_,
                random_state=idx
            )
            # Save indices
            name = f'clst-{cls}-k{k:03d}-s{idx:02d}.txt'
            print(f'  Writing file {path+name}')
            np.savetxt(path+name, sample.index, fmt='%d')

    return kmeans, cluster_metrics
````

## Evaluate clustering using logistic regression

````python
import numpy as np
from sklearn.metrics import precision_score, recall_score, f1_score

def calc_thresholds(y_label, y_proba, n_thresholds=200):
    thresholds = np.linspace(0.0, 1.0, n_thresholds)
    '''
    Calculate the precision, recall, and f1 over a range of threshold values.
    '''
    results = []
    for t in thresholds:
        y_pred = (y_proba >= t).astype(int)
        
        # Handle case where model predicts no positives
        if y_pred.sum() == 0:
            precision = 0.0
            recall = 0.0
            f1 = 0.0
        else:
            precision = precision_score(y_label, y_pred, zero_division=0)
            recall = recall_score(y_label, y_pred, zero_division=0)
            f1 = f1_score(y_label, y_pred, zero_division=0)
            results.append((t, precision, recall, f1))
    
    return np.array(results)
````

````python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import log_loss, f1_score, roc_auc_score,average_precision_score
def lr_evaluate(X_train, y_train, X_eval, y_eval):
    '''
    Evaluate the performance of a logistic regression classifer based on train and eval sets
    '''
    lr = LogisticRegression(
        solver='lbfgs',
        max_iter=1000,
        random_state=42,
    )
    lr.fit(X_train, y_train)

    y_prob = lr.predict_proba(X_eval)[:, 1]
    y_pred = (y_prob >= 0.5).astype(int)

    # Calculate the metrics at the optimum threshold
    results = calc_thresholds(y_eval, y_prob, n_thresholds=200)
    best_idx = np.argmax(results[:, 3])
    best_threshold, best_precision, best_recall, best_f1 = results[best_idx]
    
    return {
        'loss': log_loss(y_eval, y_prob),
        'f1': f1_score(y_eval, y_pred),
        'pr_auc': roc_auc_score(y_eval,y_prob),
        'best_threshold': best_threshold,
        'best_precision': best_precision,
        'best_recall': best_recall,
        'best_f1': best_f1
    }
````

````python
import json, os
METRICS_FILE = '..//results//clustering//lr_metrics.json'

# Define helper functions to read and write LR metrics from a json file
def load_metrics():
    if os.path.exists(METRICS_FILE):
        with open(METRICS_FILE, 'r') as f:
            return json.load(f)
    return {}

def save_metrics(metrics_dict):
    with open(METRICS_FILE, 'w') as f:
        json.dump(metrics_dict, f, indent=2)
````

````python
# Summarse the LR metrics
def summarise_metrics(metrics_dict):
    summary = {}
    for key, scenario in metrics_dict.items():
        metrics = scenario["metrics"]
        pr_aucs = [m["pr_auc"] for m in metrics]
        f1s = [m["best_f1"] for m in metrics]
        recalls = [m["best_recall"] for m in metrics]
        precisions = [m["best_precision"] for m in metrics]

        summary[key] = {
            "n_samples": len(metrics),
            "pr_auc_mean": mean(pr_aucs),
            "pr_auc_std": stdev(pr_aucs),
            "f1_mean": mean(f1s),
            "f1_std": stdev(f1s),
            "recall_mean": mean(recalls),
            "precision_mean": mean(precisions),
        }
    return summary
````
