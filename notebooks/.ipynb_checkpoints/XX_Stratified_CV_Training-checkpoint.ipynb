**Purpose of this notebook**  
This notebook implement stratified cross-validation training. The main training loop iterates through K folds for every augmentation level, for every scenario. Results for each fold are saved, then colated and evaluated in a separate evaluation notebook. Real training data, synthetic data, model hyperparameters and sample indices for each scenario were precompueted in notebooks

<img src="trainin_loop.png" width="350" >

## Reload the training data

```python
import pandas as pd

# Read the training and test sets
X_train = pd.read_parquet('..\\data\\processed\\X_train.parquet', engine='pyarrow')
y_train = pd.read_parquet('..\\data\\processed\\y_train.parquet', engine='pyarrow')['TARGET']

print(X_train.shape, y_train.shape)
```

```python
X_train
```

## Generate training folds


K of 5 selected as a reasonable balance between reducing variance amongst folds and having sufficient minority samples in the smallest most imbalanced scenario. K is held constant to avoid a possible confounding factor.

```python
import os
from sklearn.model_selection import StratifiedKFold
import numpy as np

# Only the tier name, IRs and number of folds are required right now. Store size for later use.
scenarios = {
    'large': {'size': 89368, 'irs': [3, 9, 27, 81, 243], 'k_folds': 5},
    'medium': {'size': 44684, 'irs': [1, 3, 9, 27, 81], 'k_folds': 5},
    'small': {'size': 22342, 'irs': [1, 3, 9, 27, 81], 'k_folds': 5}
}
n_bootstraps = 10                         # Likely only 1 or 2 will be used

# TESTING
scenarios = {
    'small': {'size': 22342, 'irs': [1, 3], 'k_folds': 5}
}

# For each scenario, bootstrap, and fold, generate train and eval indices
for tier, params in scenarios.items():
    k_folds = params['k_folds']
    for ir in params['irs']:
        for b in range(n_bootstraps):

            # Read the scenario indices
            scenario_name = f'{tier}-ir{ir:03d}-b{b:02d}'
            fullpath = os.path.join(
                '..\\scenarios', scenario_name, 
                'training-'+scenario_name+'.txt')
            scenario_indices = np.loadtxt(fullpath, dtype=int).tolist()
            
            # Select the scenario from the full train DataFrame
            X_scenario = X_train.loc[scenario_indices]
            y_scenario = y_train.loc[scenario_indices]

            # Generate the fold indices. Ensure different random state is used per bootstrap
            random_state = 42 + b
            skf = StratifiedKFold(n_splits=k_folds, shuffle=True, random_state=random_state)
            folds = list(skf.split(X_scenario, y_scenario))

            # Lookup the sample Ids using the fold indices and save them as
            # - fold-k-train-ids.txt
            # - fold-k-eval-ids.txt
            for k, f in enumerate(folds):
                fold_train_ids = (X_scenario.iloc[f[0]].index)
                fold_eval_ids = (X_scenario.iloc[f[1]].index)
                #print('Train fold indices:', train_fold_ids[:5])
                #print('Eval fold indices:', eval_fold_ids[:5])

                fullpath = os.path.join(
                    '..\\scenarios', scenario_name, 'folds',
                    'fold-'+str(k)+'-train-ids.txt')
                np.savetxt(fullpath, fold_train_ids, fmt='%d')
                fullpath = os.path.join(
                    '..\\scenarios', scenario_name, 'folds',
                    'fold-'+str(k)+'-eval-ids.txt')
                np.savetxt(fullpath, fold_eval_ids, fmt='%d')
```
